```python
# В первой ячейке ноутбука
from openbb import obb
import pandas as pd
import matplotlib.pyplot as plt

# Настройка отображения
plt.style.use('seaborn-v0_8')
%matplotlib inline
```

```python
# Получение данных
aapl_data = obb.equity.price.historical("AAPL", start_date="2020-01-01")
df = aapl_data.to_df()

# Базовая информация
print(f"Данные с {df.index.min()} по {df.index.max()}")
print(f"Колонки: {list(df.columns)}")
df.describe()
```

```python
# График цен
plt.figure(figsize=(12, 6))
plt.plot(df.index, df['close'], label='Цена закрытия AAPL')
plt.title('История цен Apple')
plt.xlabel('Дата')
plt.ylabel('Цена ($)')
plt.legend()
plt.grid(True)
plt.show()
```

```python
# Экономические данные - например, инфляция в США
inflation = obb.economy.cpi(country='united_states')
inflation_df = inflation.to_df()
print(inflation_df.head())

# Данные по криптовалюте
btc_data = obb.crypto.price.historical("BTC-USD", start_date="2023-01-01")
print(f"BTC данные: {btc_data.to_df().shape}")
```
